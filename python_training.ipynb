```python
r = int(input("Enter number of rats: "))
unit = int(input("Enter amount of food each rat eats: "))
n = int(input("Enter number of houses: "))
arr = list(map(int, input("Enter food in each house: ").split()))
if arr is None:
    print(-1)
else:
    required = r * unit
    total = 0
    houses = 0
    for i in range(n):
        total = total + arr[i]
        houses = houses + 1
        if total >= required:
            break
    if total < required:
        print(0)
    else:
        print("Minimum number of houses:", houses)
```

```python
print("SUM OF THE DIGITS")
text = input("Enter a number: ")
if text.isdigit():
    total = 0
    for digit in text:
        total = total + int(digit)
    print("Sum of digits:", total)
else:
     print("Please enter numbers only.")
```

```python
print("FIND THE LARGEST DIGIT IN A NUMBER")
n = int(input("Enter the number: "))
a = 0
while n > 0:
    r = n % 10
    if r > a:
        a = r
    n = n // 10
print("Largest digit:", a)
```

```python
print("FIND STRING PALINDROME")
n=int(input("Enter a number:"))
a=n
t=0
while(n>0):
  r=n%10
  t=(t*10)+r
  n=n//10
if(a==t):
  print("Palindrome")
else:
  print("Not Palindrome")
```
